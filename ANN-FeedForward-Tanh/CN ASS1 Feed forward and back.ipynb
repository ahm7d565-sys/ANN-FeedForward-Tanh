```python
import numpy as np
```

```python
np.random.seed(1)
```

```python
x1 = 0.05
x2 = 0.10
X = np.array([x1, x2])
```

```python
W1 = np.random.uniform(-0.5, 0.5, (2, 2))
W2 = np.random.uniform(-0.5, 0.5, (2, 2))
```

```python
b1 = 0.5
b2 = 0.7
```

```python
def tanh(x):
    return np.tanh(x)
```

```python
net_h = np.dot(X, W1) + b1
out_h = tanh(net_h)
```

```python
net_o = np.dot(out_h, W2) + b2
out_o = tanh(net_o)
```

```python
print("W1 =\n", W1)
print("\nW2 =\n", W2)

print("\nHidden Layer Output = ", out_h)
print("\nFinal Network Output = ", out_o)
```
